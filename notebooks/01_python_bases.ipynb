## Conteneurs

Python propose de nombreux types de conteneurs efficaces, dans lesquels on peut stocker des collections d'objets.

### Listes

Une liste est une collection **ordonnée** d'objets, qui peuvent être de types différents. Par exemple :

````python
couleurs = ['rouge', 'bleu', 'vert', 'noir', 'blanc']
type(couleurs)
````

**Indexation** : accéder à un élément particulier de la liste :

````python
couleurs[2]
````

On peut compter à partir de la fin avec des indices négatifs :

````python
couleurs[-1]
````

````python
couleurs[-2]
````

> ⚠️ **Attention**
>
> **L'indexation commence à 0** (comme en C), et non à 1 (comme en Fortran ou en Matlab) !


**Découpage** (*slicing*) : obtenir des sous-listes d'éléments régulièrement espacés :

````python
couleurs
````

````python
couleurs[2:4]
````

> ⚠️ **Attention**
>
> Notez que `couleurs[debut:fin]` contient les éléments d'indice `i` tels que `debut <= i < fin` (`i` allant de `debut` à `fin-1`). Par conséquent, `couleurs[debut:fin]` contient `(fin - debut)` éléments.


**Syntaxe du découpage** : `couleurs[debut:fin:pas]`

Tous les paramètres du découpage sont facultatifs :

````python
couleurs
````

````python
couleurs[3:]
````

````python
couleurs[:3]
````

````python
couleurs[::2]
````

Les listes sont des objets **modifiables** (*mutable*) :

````python
couleurs[0] = 'jaune'
couleurs
````

````python
couleurs[2:4] = ['gris', 'violet']
couleurs
````

> 📝 **Note**
>
> Les éléments d'une liste peuvent avoir des types différents :
>

````python
couleurs = [3, -200, 'bonjour']
couleurs
````

>

````python
couleurs[1], couleurs[2]
````

>
> Pour des collections de données numériques de même type, il est souvent **bien plus efficace** d'utiliser le type `array` du module `numpy`. Un tableau NumPy est un bloc de mémoire contenant des éléments de taille fixe. Avec les tableaux NumPy, les opérations sont plus rapides car les éléments sont régulièrement espacés en mémoire et la plupart des opérations passent par des fonctions C spécialisées plutôt que par des boucles Python. Nous y reviendrons dans le carnet suivant.


Python offre de nombreuses fonctions pour modifier ou interroger les listes. Voici quelques exemples ; pour plus de détails, voir <https://docs.python.org/fr/3/tutorial/datastructures.html#more-on-lists>.

Ajouter et retirer des éléments :

````python
couleurs = ['rouge', 'bleu', 'vert', 'noir', 'blanc']
couleurs.append('rose')
couleurs
````

````python
couleurs.pop() # retire et renvoie le dernier élément
````

````python
couleurs
````

````python
couleurs.extend(['rose', 'violet']) # étend la liste sur place
couleurs
````

````python
couleurs = couleurs[:-2]
couleurs
````

Inverser :

````python
rcouleurs = couleurs[::-1]
rcouleurs
````

````python
rcouleurs2 = list(couleurs) # nouvel objet : une copie de couleurs, à un autre emplacement mémoire
rcouleurs2
````

````python
rcouleurs2.reverse() # sur place ; inverser rcouleurs2 ne modifie pas couleurs
rcouleurs2
````

Concaténer et répéter des listes :

````python
rcouleurs + couleurs
````

````python
rcouleurs * 2
````

**Trier :**

````python
sorted(rcouleurs) # nouvel objet
````

````python
rcouleurs
````

````python
rcouleurs.sort()  # sur place
rcouleurs
````

> **Méthodes et programmation orientée objet**
>
> La notation `rcouleurs.methode()` (par exemple `rcouleurs.append(3)` ou `couleurs.pop()`) est notre premier exemple de programmation orientée objet (POO). Puisque `rcouleurs` est une `list`, cet objet possède des *méthodes*, qu'on appelle avec la notation **.** (le point). Il n'est pas nécessaire d'en savoir plus sur la POO que cette notation pour suivre ce cours.

### Chaînes de caractères

Différentes syntaxes sont possibles (apostrophes simples, guillemets doubles ou triples) :

````python
s = 'Bonjour, comment allez-vous ?'
s = "Salut, ça va ?"
s = '''Bonjour,
       comment allez-vous ?'''    # tripler les guillemets permet à la chaîne
                                  # de s'étendre sur plusieurs lignes
s = """Salut,
ça va ?"""
````

Cependant, si vous essayez d'exécuter ce code :

```text
'Salut, c'est moi'
```

vous obtiendrez une erreur de syntaxe. (Essayez. Pourquoi ?)

Cette erreur peut être évitée en entourant la chaîne de guillemets doubles plutôt que d'apostrophes. On peut aussi faire précéder la deuxième apostrophe d'une barre oblique inverse (*backslash*). Cette barre sert aussi à écrire le saut de ligne `\n` et la tabulation `\t`.

Les chaînes sont des collections, comme les listes. On peut donc les indexer et les découper avec la même syntaxe et les mêmes règles.

Indexation :

````python
a = "bonjour"
a[0]
````

````python
a[1]
````

````python
a[-1]
````

(Rappel : les indices négatifs correspondent à un comptage à partir de la fin.)

````python
a = "bonjour, monde !"
a[3:6] # du 3e au 6e élément (exclu) : éléments 3, 4, 5
````

````python
a[2:10:2] # Syntaxe : a[debut:fin:pas]
````

> 📝 **Note**

````python
a.replace('o', 'z', 1)
````

## L'opérateur d'affectation

La [référence du langage Python](https://docs.python.org/fr/3/reference/simple_stmts.html#assignment-statements) indique :

> Les instructions d'affectation sont utilisées pour lier (ou relier) des noms à des valeurs et pour modifier des attributs ou des éléments d'objets modifiables.

En bref, cela fonctionne ainsi (affectation simple) :

1. l'expression à droite est évaluée, et l'objet correspondant est créé ou obtenu ;
2. un **nom** à gauche est affecté (lié) à cet objet.

À retenir :

- Un même objet peut avoir plusieurs noms :

````python
a = [1, 2, 3]
b = a
a
````

````python
b
````

````python
a is b
````

````python
b[1] = 'salut !'
a
````

- pour modifier une liste *sur place*, utilisez l'indexation ou le découpage :

````python
a = [1, 2, 3]
a
````

````python
a = ['a', 'b', 'c'] # Crée un autre objet.
a
````

````python
id(a)
````

````python
a[:] = [1, 2, 3] # Modifie l'objet sur place.
a
````

````python
id(a)
````

- la notion clé ici est la différence **modifiable / immuable** :
  - les objets modifiables peuvent être changés sur place ;
  - les objets immuables ne peuvent plus être modifiés une fois créés.
